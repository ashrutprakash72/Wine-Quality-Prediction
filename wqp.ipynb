Importing the dependencies

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

Data Collection

```python
# loading the data set to a pandas data frame
wine_dataset = pd.read_csv('winequality-red.csv')
```

```python
# number of rows & columns in the dataset
wine_dataset.shape
```

```python
# first 5 rows of the dataset
wine_dataset.head()
```

```python
# checking for missing values
wine_dataset.isnull().sum()
```

Data Analysis and visualaization

```python
# statistically measures of the dataset
wine_dataset.describe()
```

```python
# number of values for each quality
sns.catplot(x = 'quality', data = wine_dataset, kind = 'count')
```

```python
 # volatile acidity vs quality
plot = plt.figure(figsize=(5,5))
sns.barplot(x='quality',y= 'volatile acidity', data = wine_dataset)
```

```python
 # citric acidity vs quality
plot = plt.figure(figsize=(5,5))
sns.barplot(x='quality',y= 'citric acid', data = wine_dataset)
```

Correlation 

 1. Positive Correlation
 2. negative correlation

```python
correlation = wine_dataset.corr()
```

```python
# constructing to heatmap to understand the correlation between the columns
plt.figure(figsize=(10,10))
sns.heatmap(correlation, cbar = True, square=True, fmt = '.If', annot = True, annot_kws={'size':8}, cmap = 'Blues')
```

Data preprocessing

```python
# separate the data and label
X = wine_dataset.drop('quality',axis=1)
```

```python
print(X)
```

label binarization

```python
Y = wine_dataset['quality'].apply(lambda y_value: 1 if y_value>=7 else 0)
```

```python
print(Y)
```

Train & Test Split

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size = 0.2, random_state=2)
```

```python
print(Y.shape, Y_train.shape, Y_test.shape)
```

Modal Training

Random ForestClassifier

```python
model = RandomForestClassifier()
```

```python
model.fit(X_train, Y_train)
```

Model Evaluation 

Accuracy Score

```python
# accuracy on test data

X_test_prediction = model.predict(X_test)
test_data_accuracy = accuracy_score(X_test_prediction, Y_test) 
```

```python
print('Accuracy : ' , test_data_accuracy)
```


Building a Predictive System



```python
input_data = (7.3,0.65,0.0,1.2,0.065,15.0,21.0,0.9946,3.39,0.47,10.0)

#changing the input data to numpy array

input_data_as_numpy_array = np.asarray(input_data)

#reshape the data as we are predicting the label for only one instance
input_data_reshaped = input_data_as_numpy_array.reshape(1, -1)

prediction = model.predict(input_data_reshaped)
print(prediction)

if(prediction[0]==1):
    print('good quality wine')
else:
    print('bad quality wine')
```
